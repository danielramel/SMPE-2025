```python
data1 = [x*1000 for x in [139, 146, 146, 133, 134, 135,137,144,142,134,130]]
data2 = [8,11,10,3,7,4,6,5,8,2,6]
# data1 is juvenile drug arrests, data2 is number of kids killed by their parent
#make two separate line plots with the same x-axis (year) and different y-axes (one for each dataset) to avoid misleading dual y-axis.
```

```python
import matplotlib.pyplot as plt

years = list(range(1999, 2010))

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

ax1.plot(years, data1, marker='o', label='Juvenile Drug Arrests')
ax1.set_ylabel('Number of Arrests')
ax1.set_title('Juvenile Drug Arrests by Year')
ax1.grid(True)
ax1.set_ylim(0, max(data1) + 5000)

ax2.plot(years, data2, marker='o', color='orange', label='Kids Killed by Parent')
ax2.set_xlabel('Year')
ax2.set_ylabel('Number of Deaths')
ax2.set_title('Kids Killed by Parent by Year')
ax2.grid(True)
ax2.set_ylim(0, max(data2) + 5)

plt.tight_layout()
fig.savefig('figure1_good.png')
plt.show()
```
